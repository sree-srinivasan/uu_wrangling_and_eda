## Q1 (Which country gives the most money in total?)

```python
df.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending = False)
```

### Answer

### Answer

## Q3 (Which country gives the largest gifts on average?)

```python
df.groupby('Country of Giftor')['Foreign Gift Amount'].mean().sort_values(ascending = False)
```

### Answer 
Bermuda has the highest average at 7.688837e+06

## Q4 (Which institution receive the most money in total?)

## Top 20 instituation by total amount of gifts

```python
top20 = df.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values(ascending = False).head(20)
sns.barplot(x = top20.values, y = top20.index)
plt.xlabel('Total Foreign Gift Amount')
plt.ylabel('Institutions')
plt.title('Top 20 Instituations by Total Amount of Gifts ($)')
plt.show()
```

## Q6 (What is the average amount each university receives? Median amount? If they are different why?)

```python
uni = df.groupby('Institution Name')['Foreign Gift Amount'].sum()
uni_mean = uni.mean()
uni_median = uni.median()

print(uni_mean)
print(uni_median)
```

### Answer
Mean gift amount is 588232.7172672832. Median gift amount is 94615.0. Mean of the institution total is 52202878.974842764. Median of the institution toal is 6486791.5. Numbers are going to be really different as there probably was (actually there is) people who donated a lot more on an invidiual basis so there will be a big difference. 

```python
sns.histplot(uni, bins = 100, log_scale = True)
plt.axvline(uni_mean, color = 'black')
plt.axvline(uni_median, color = 'grey')
plt.xlabel('Foreign Gift Amount')
plt.ylabel('Count')
plt.title('Distribution of Individual Foreign Gift Amounts')
```

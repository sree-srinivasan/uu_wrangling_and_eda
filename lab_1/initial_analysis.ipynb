# Q & A 



## Q1 (Which country gives the most money in total?)

```python
df.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending = False)
```

### Answer 
Qatar is the country that gives the most money at 2706240869

## Q2 (Which country initiates the most gifts by count?)

```python
df.groupby('Country of Giftor').size().sort_values(ascending = False)
```

### Answer 
England has the most amount of gifts as 3655

## Q3 (Which country gives the largest gifts on average?)

```python
df.groupby('Country of Giftor')['Foreign Gift Amount'].mean().sort_values(ascending = False)
```

### Answer 
Bermuda has the highest average at 7.688837e+06

## Q4 (Which institution receive the most money in total?)

```python
df.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values(ascending = False)
```

### Answer 
Carnegie Mellon University had the most amount of money at 1477922504

# Executive Summary
As a whole, UVA has received about $22 million in foreign gifts over this dataset that was given to us. This puts them at 101 out of 318 U.S. academic institutions, capturing about 0.13% of that national foreign giving market.

When specifically looking at Virginia public universities, they're trailing third when compared to VCU and GMU. That really comes from UVA's lack of focus towards countries that give a lot more, like Qatar and Saudi Arabia, and also focusing on national market nations that already fund Virginia universities.

The recommendation from this analysis is to focus on those sorts of countries, like Qatar and Saudi Arabia, as well as more nationally established countries, through a channel that's been really working well for UVA, which is contracts rather than real estate or monetary gifts. 

# Analysis
At a national level, it's pretty highly concentrated:
1. Qatar is the largest source.
2. England.
3. China.
4. Saudi Arabia.
5. Bermuda.

Bermuda and Qatar also have the largest average gifts. Carnegie Mellon is the one that receives the most amount of money, while UCLA receives the most amount of gifts by individual count.

Looking at UVA specifically, those giving splits are pretty much even. Actually, not really, but they're almost even: the majority is in contracts, and some are monetary gifts from mostly countries that are not well established or well known in the nation, like Bangladesh or Tanzania, etc.

# Ways for New Sources

The best way for UVA is to go after a country like Qatar or Saudi Arabia. These are top-tier national givers with also large average gift sizes, and they already fund Virginia universities like VCU and GMU. UVA is probably better off at just copying VCU and what they're doing first with guitar, and then they can adapt to whatever works for UVA specifically, whether that be through research or any other initiative. The other larger nations, like China, Canada, and the UAE, are definitely untapped markets that would fit strongly for UVA they can continue to do research contracts with these larger nations, or they're probably going to be able to even get one-off donations and not have anything in return, really. This way, UVA has a double-edged approach:
- Work for the money you get.
- Target countries that can afford to spend some money as an addition to their budget or negligible side of their budget and their nation's GDP.

UVA should also try to get an internal question answered about why some of these nations that are giving higher-value contracts (for example, Qatar or Bermuda) are missing from their own portfolio of donations. That's a really good way to bridge that gap and offer something of value to the other nations so that they get more donations as well

# Conclusion
UVA's gap in foreign giving is really just a problem that looks like a market access problem rather than a capacity issue. By just being around the area, I know it's not a capacity issue for sure, but looking at the data specifically, the university does a lot of research, and all their contracts are as such. All these are partnerships that are based off of input that the university is doing to get some output, which is the money or the donation.

As long as UVA can start to prioritize a better source of donations in return for work or contracts, that would be the best way to raise as much money as possible. It would also leverage the prestige and the public university status, or the ranking of where UVA stands, as a way to target more established countries to get those donations without having to do any extra work. That way, they're able to capture both nations that are willing to spend a lot of money in exchange for something and countries that have enough budget to just help out a university. That could definitely bridge the gap between VCU and UVA and help them get to the top of Virginia's donations or giving. It can also put them higher up in the ranking and get closer to the Carnegie Mellon level.
